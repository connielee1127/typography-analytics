```python
industry_counts = df["industry"].value_counts()

for threshold in [1000, 2500, 5000, 10000, 25000]:
    print(
        f"{threshold:,}: "
        f"{(industry_counts >= threshold).sum()} industries"
    )
```

```python
eligible = (
    industry_counts[industry_counts >= 1000]
    .rename("company_count")
    .reset_index()
    .rename(columns={"index": "industry"})
)

eligible
```

```python
pd.set_option("display.max_rows", None)
eligible
```

```python
mapping = eligible.copy()

mapping["industry_group"] = ""
mapping["review_status"] = "needs_review"

mapping.to_csv(
    "../data/reference/industry_mapping.csv",
    index=False
)
```

Taking a larger sample of data to observe the proportion of company websites that have social media/directory domains (not actual company domains)

```python
from urllib.parse import urlparse

sample = df.sample(n=10_000, random_state=42)

domains = sample["website"].dropna().reset_index(drop=True)

domains
```

```python
print(domains.value_counts().head(50))
```

```python
exclude_domains = [
    "yelp.com",
    "facebook.com",
    "instagram.com",
    "linkedin.com",
    "linktr.ee",
    "google.com",
    "bit.ly",
    "youtube.com",
]
```

```python
excluded_count = domains.isin(exclude_domains).sum()

print("Excluded:", excluded_count)
print("Remaining:", len(domains) - excluded_count)
print("Percent excluded:", excluded_count / len(domains) * 100)
```
