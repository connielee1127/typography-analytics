```python
import pandas as pd
```

```python
us_companies = pd.read_csv("../data/processed/us_companies.csv")
us_companies.head()
```

```python
mapping = pd.read_csv("../data/reference/industry_mapping.csv")
mapping.head()
```

```python
merged = us_companies.merge(
    mapping[["industry", "industry_group"]],
    on="industry",
    how="left"    
)
```

```python
merged.head()
```

```python
merged["industry_group"].isna().sum()
```

```python
merged["industry_group"].value_counts()
```

```python
mapping["industry_group"].value_counts()
```

```python
len(merged)
```

```python
nan_removed = merged[merged["industry_group"].notna()]
sample_population = nan_removed[nan_removed["industry_group"] != "Ambiguous"]
sample_population
```

```python
sample_population["industry_group"].value_counts()
```

```python
len(sample_population)
```

```python
sample_population["industry_group"].isna().sum()
```

```python
sample_population.to_csv(
    "../data/processed/sampling_population.csv",
    index=False
)
```

Created sampling_population.csv - US-only company data with companies from the top 19 industry groups; filtered out industries classified as Ambiguous
