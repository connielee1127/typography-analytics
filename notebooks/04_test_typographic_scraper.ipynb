```python
import asyncio
import pandas as pd

import sys
sys.path.append("..")

from src.url_validator import validate_website
from src.typography_scraper import load_page
```

```python
df = pd.read_csv("../data/processed/sampling_population.csv")
df
```

```python
sample = df.sample(40, random_state=42)
sample
```

```python
url_access = sample["website"].apply(validate_website)
access_results = pd.DataFrame(url_access.tolist())
```

```python
success = access_results[access_results["website_type"] == "company_website"]
success
```

```python
len(success)
```

```python
typography_results = await load_page(success["final_url"].iloc[0])

typography_results
```

```python
urls = success["final_url"].tolist()

typography_results_all = await asyncio.gather(
    *(load_page(url) for url in urls)
)

typography_results_all
```
