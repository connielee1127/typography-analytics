```python
import pandas as pd
import sys
sys.path.append("..")
from src.url_validator import normalize_url, check_url_reachability, validate_website
```

```python
sample_big = pd.read_csv(
    "../data/processed/us_companies.csv",
    nrows=1000
)

sample_1 = sample_big.sample(100, random_state=42)
sample_1
```

```python
sample_1["website_normalized"] = sample_1["website"].apply(normalize_url)
sample_1[["name", "website", "website_normalized"]]
```

```python
sample_1["website_normalized"].isna().sum()
```

```python
sample_1[
    sample_1["website"] != sample_1["website_normalized"]
][["name", "website", "website_normalized"]]
```

URL format observation: In a random sample of 100 companies, all website values were stored as domain-only strings without an HTTP/HTTPS scheme. normalize_url() therefore added https:// to all sampled URLs, with no missing outputs.

```python
sample_big["website"].str.contains("://", regex=False).sum()
```

```python
sample_big[
    sample_big["website"].str.contains("/", regex=False)
][["name", "website"]]
```

The dataset's website field contains bare domains rather than complete URLs, so normalization only needs to add an HTTPS scheme and handle missing/invalid values.

```python
sample_2 = sample_1.sample(25, random_state=42).copy()
sample_2
```

```python
reachability_results = sample_2["website_normalized"].apply(check_url_reachability)
reachability_results
```

```python
reachability_df = pd.DataFrame(reachability_results.tolist())
reachability_df
```

```python
reachability_df["accessible"].value_counts()
```

```python
reachability_df[
    reachability_df["accessible"] == False
][[
    "http_status",
    "content_type",
    "failure_reason"
]]
```

```python
sample_2 = pd.concat(
    [sample_2.reset_index(drop=True), reachability_df],
    axis=1
)
sample_2
```
